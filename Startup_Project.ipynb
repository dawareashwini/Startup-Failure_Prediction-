```python
df_Founder_Burnout_Flag = df[['Economic_Climate', 'Founder_Age', 'Founder_Experience_Years', 'Industry', 'Funding_Stage', 'Work_Mode', 'Team_Size', 'Startup_Age_Months', 'Weekly_Work_Hours', 'Monthly_Revenue_Growth_Percent', 'Runway_Months_Remaining', 'Product_Market_Fit_Score', 'Employee_Turnover_Percent', 'Work_Life_Balance_Score', 'Seeks_Mental_Health_Support', 'Founder_Burnout_Flag']]
```

```python
X = df_Founder_Burnout_Flag.drop(columns=['Founder_Burnout_Flag'])
y = df_Founder_Burnout_Flag['Founder_Burnout_Flag']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, f1_score, confusion_matrix, accuracy_score
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import pandas as pd

transformer = ColumnTransformer(
    transformers=[
        ('tf1', OneHotEncoder(sparse_output=False), ['Economic_Climate', 'Industry', 'Funding_Stage', 'Work_Mode', 'Seeks_Mental_Health_Support']),
        ('tf2', StandardScaler(), ['Founder_Age', 'Founder_Experience_Years', 'Team_Size', 'Startup_Age_Months', 'Weekly_Work_Hours', 'Monthly_Revenue_Growth_Percent','Runway_Months_Remaining','Product_Market_Fit_Score','Employee_Turnover_Percent','Work_Life_Balance_Score'])
    ],
    remainder='drop'
)
```

```python
# final model is LogisticRegression, so we build there pipeline for automation

from sklearn.pipeline import Pipeline

pipe = Pipeline(
    [
        ('Preprocessing', transformer),
        ('lr', LogisticRegression())
    ]
)

pipe.fit(x_train, y_train)
y_pred = pipe.predict(x_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("F1 score:", f1_score(y_test, y_pred, average='weighted'))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
